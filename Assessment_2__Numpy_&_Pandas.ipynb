```python
#Q3

import numpy as np
def describe_array_stats(data):
    # Perform calculations
    arr_mean = np.mean(data)
    arr_median = np.median(data)
    arr_min = np.min(data)
    arr_max = np.max(data)
    arr_std = np.std(data)

    # Print results with descriptions
    print("NumPy Statistical operations")
    print(f"Mean: {arr_mean}")
    print(f"Median: {arr_median}")
    print(f"Minimum: {arr_min}")
    print(f"Maximum: {arr_max}")
    print(f"Standard Deviation: {arr_std:.4f}")

# Given array
arr = np.array([12, 15, 18, 20, 25, 30])

# Call the function
describe_array_stats(arr)


```

```python
#D
import pandas as pd
import numpy as np

# 1. Restore df_std_per with student performance data
file_path = "/content/drive/MyDrive/DSA/Assessment/Assessment 2  Numpy & Pandas/student_performance.csv"
df_std_per = pd.read_csv(file_path)

# Clean missing values
numerical_cols = df_std_per.select_dtypes(include=['float64', 'int64']).columns
for col in numerical_cols:
    df_std_per[col] = df_std_per[col].fillna(df_std_per[col].mean())

# Re-engineer required features
subject_cols = ['Python_Score', 'SQL_Score', 'Pandas_Score', 'NumPy_Score']
df_std_per['Average_Score'] = df_std_per[subject_cols].mean(axis=1)
df_std_per['Pass_Status'] = df_std_per['Average_Score'].apply(lambda score: 'Pass' if score >= 50 else 'Fail')

# Standardize department names
df_std_per['Department'] = df_std_per['Department'].str.strip().str.title()

# 2. Generate the department summary
df_dept_summary = df_std_per.groupby('Department').agg(
    Total_Students=('Student_ID', 'count'),
    Average_Score=('Average_Score', 'mean'),
    Average_Attendance=('Attendance', 'mean'),
    Pass_Count=('Pass_Status', lambda x: (x == 'Pass').sum()),
    Fail_Count=('Pass_Status', lambda x: (x == 'Fail').sum())
).reset_index()

# Round numerical metrics
df_dept_summary['Average_Score'] = df_dept_summary['Average_Score'].round(2)
df_dept_summary['Average_Attendance'] = df_dept_summary['Average_Attendance'].round(2)

# Display the final summary DataFrame
display(df_dept_summary)
```

```python
#17


import pandas as pd
file_path= "/content/drive/MyDrive/DSA/Assessment/Assessment 2  Numpy & Pandas/sales.csv"
df_sale=pd.read_csv(file_path)
df_sale
```

```python
df_sale['Order_Date'] = pd.to_datetime(df_sale['Order_Date'], format='mixed', dayfirst=True)
print("Order_Date converted to:", df_sale['Order_Date'].dtype)
```

```python
missing_sales = df_sale.isnull().sum()
print("Missing Values Count per Column in df_sale:")
print(missing_sales[missing_sales > 0])
```

```python
import pandas as pd

# 1. Calculate general metrics
total_sales = df_sale['Sales'].sum()
total_profit = df_sale['Profit'].sum()
avg_sales = df_sale['Sales'].mean()

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Average Sales: ${avg_sales:,.2f}\n")

# 2. Top 5 products based on sales
top_products = df_sale.groupby('Product')['Sales'].sum().reset_index()
top_products_sorted = top_products.sort_values(by='Sales', ascending=False).head(5)
print("--- Top 5 Products by Sales ---")
display(top_products_sorted)
print()

# 3. Region generating the highest sales
region_sales = df_sale.groupby('Region')['Sales'].sum().reset_index()
best_region = region_sales.sort_values(by='Sales', ascending=False).iloc[0]
print(f"Region with Highest Sales: {best_region['Region']} (${best_region['Sales']:,.2f})")
```

```python
import pandas as pd
import numpy as np

# Convert columns to numeric types to fix aggregation errors
df_sale['Sales'] = pd.to_numeric(df_sale['Sales'], errors='coerce')
df_sale['Profit'] = pd.to_numeric(df_sale['Profit'], errors='coerce')
df_sale['Discount'] = pd.to_numeric(df_sale['Discount'], errors='coerce')

# --- Part C: Category Analysis ---
category_summary = df_sale.groupby('Category').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Average_Discount=('Discount', 'mean')
).reset_index()

# Sort by Total Sales in descending order
category_summary = category_summary.sort_values(by='Total_Sales', ascending=False)

print("--- Category Analysis Summary ---")
display(category_summary)

# --- Part D: Extract Data for Business Insights ---
# 1. Most profitable category
most_profitable_cat = category_summary.sort_values(by='Total_Profit', ascending=False).iloc[0]

# 2. Region with highest sales
region_sales = df_sale.groupby('Region')['Sales'].sum().reset_index()
best_region = region_sales.sort_values(by='Sales', ascending=False).iloc[0]

# 3. Product with highest sales
product_sales = df_sale.groupby('Product')['Sales'].sum().reset_index()
best_product = product_sales.sort_values(by='Sales', ascending=False).iloc[0]

print("\n--- Key Business Findings ---")
print(f"1. Most Profitable Category: {most_profitable_cat['Category']} (${most_profitable_cat['Total_Profit']:,.2f})")
print(f"2. Region with Highest Sales: {best_region['Region']} (${best_region['Sales']:,.2f})")
print(f"3. Product with Highest Sales: {best_product['Product']} (${best_product['Sales']:,.2f})")
```
